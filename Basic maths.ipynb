## Reorganising Array

```python
before=np.array([[1,2,3,4],[5,6,7,8]])
print(before)

after=before.reshape((8,1))
print(after)

#after=before.reshape((8,1))
# why we write 8,1 within 2 brackets?
#We use two sets of parentheses because the outer pair calls the .reshape() method,
#  while the inner pair wraps 8, 1 into a single Python tuple representing the array's new shape
```

```python
#Vertically stacking vectors
v1=np.array([[1,2,3,4]])
v2=np.array([[5,6,7,8]])
v3=np.vstack([v1,v2,v1,v2])
print(v3)
```

```python
#Horizontally stcaking vectors
h1=np.array([[1,2,3,4]])
h2=np.array([[5,6,7,8]])
h3=np.hstack((h1,h2))
print(h3)

h4=np.ones((2,4))
h5=np.zeros((2,2))
h6=np.hstack([h4,h5])
print(h6)
```

# Boolean masking and Advanced indexing

```python
s>18
```

```python
np.any(s>18,axis=0) 

# Columns 1–4 ([1, 22], [2, 55], [3, 87], [4, 66]): True — The bottom numbers (22, 55, 87, 66) are all > 18.
# Column 5 ([5, 11]): False — Neither 5 nor 11 is > 18.
```

```python
np.any(s>18,axis=1)

#Row 1 ([1, 2, 3, ..., 15]): False — None of the 15 numbers in the top row are greater than 18 (the highest is 15).
#Row 2 ([22, 55, 87, ..., 7]): True — At least one number in the bottom row is greater than 18 (in fact, eight of them are: 22, 55, 87, 66, 47, 86, 78, 36).
```
